```python
import cv2
import matplotlib.pyplot as plt

# 1. Load the image from your uploaded file
image_path = "portrait1.jpg"  # Replace with your filename or /content/filename
img_bgr = cv2.imread(image_path)

if img_bgr is None:
    raise FileNotFoundError(f"Could not load image at {image_path}. Check the filename!")

# Convert BGR to RGB for Matplotlib
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# 2. Apply Gaussian Filters with varying sigma and kernel sizes
# (ksize must be positive and odd)
blur_light = cv2.GaussianBlur(img_rgb, ksize=(5, 5), sigmaX=1.0)
blur_medium = cv2.GaussianBlur(img_rgb, ksize=(11, 11), sigmaX=2.0)
blur_heavy = cv2.GaussianBlur(img_rgb, ksize=(21, 21), sigmaX=5.0)

# 3. Display original vs blurred versions side-by-side
fig, axes = plt.subplots(1, 4, figsize=(18, 5))

axes[0].imshow(img_rgb)
axes[0].set_title("Original Image", fontsize=12)
axes[0].axis('off')

axes[1].imshow(blur_light)
axes[1].set_title("5x5 Kernel ($\sigma = 1.0$)\nMild Denoising", fontsize=12)
axes[1].axis('off')

axes[2].imshow(blur_medium)
axes[2].set_title("11x11 Kernel ($\sigma = 2.0$)\nMedium Smoothing", fontsize=12)
axes[2].axis('off')

axes[3].imshow(blur_heavy)
axes[3].set_title("21x21 Kernel ($\sigma = 5.0$)\nHeavy Blur", fontsize=12)
axes[3].axis('off')

plt.tight_layout()
plt.show()
```
